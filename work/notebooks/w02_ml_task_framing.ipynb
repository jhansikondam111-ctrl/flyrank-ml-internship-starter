## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as "target customers who visited the website more than 5 times" is simple, but it may not work well for every customer.

Customer behaviour depends on multiple factors such as previous purchases, website visits and engagement.

ML can learn patterns from multiple features and predict which customers are more likely to purchase.

Therefore, ML is more flexible and data-driven than using a single fixed rule.

The prediction can support a real action such as sending targeted product recommendations or promotional content to customers who are likely to purchase.
